```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import seaborn as sns
```

```python
file_path = "./(202510)VinUni Charging Data.xlsx"
```

```python
all_sheets = pd.read_excel(file_path, sheet_name=None)
```

```python
sheet_names = list(all_sheets.keys())
```

```python
df_C = all_sheets[sheet_names[1]]
df_J = all_sheets[sheet_names[2]]
df_all = all_sheets[sheet_names[3]]
```

```python
df_all
```

```python
df_all1 = df_all[~df_all['Week'].isin(['Testing'])]
df_all1 = df_all1[~df_all1['Weekday'].isin(['Sat', 'Sun'])]
df_all1
```

```python
df_C1 = df_C[~df_C['Week'].isin(['Testing'])]
df_C1 = df_C1[~df_C1['Weekday'].isin(['Sat', 'Sun'])]
df_C1
```

```python
df_J1 = df_J[~df_J['Week'].isin(["Testing"])]
df_J1 = df_J1[~df_J1['Weekday'].isin(['Sat', 'Sun'])]
df_J1
```

```python
df_combined = df_all1

df_combined['Date'] = pd.to_datetime(df_combined['Date'])
df_filtered = df_combined[df_combined['Total cars'] > 0].copy()

df_daily_traffic = (
    df_filtered.groupby(['Date', 'Time'])[['Total EV cars', 'Gasoline cars', 'Total cars']]
    .sum()
    .reset_index()
)

df_daily_traffic.sort_values(by='Date', inplace=True)
```

```python
df_morning = df_daily_traffic[df_daily_traffic['Time'] == '09:00–10:00'].copy().reset_index(drop=True)
df_afternoon = df_daily_traffic[df_daily_traffic['Time'] == '14:00–15:00'].copy().reset_index(drop=True)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 7), sharey=True)
plt.style.use('seaborn-v0_8-whitegrid')

plot_lines = ['Total EV cars', 'Gasoline cars', 'Total cars']
colors = {'Total EV cars': 'skyblue', 'Gasoline cars': 'orange', 'Total cars': 'darkblue'}
date_formatter = mdates.DateFormatter('%m-%d')

def plot_traffic(df, ax, title_text, legend_loc):
    """Plots traffic data for a specific time slot."""
    df.set_index('Date')[plot_lines].plot(
        kind='line', 
        ax=ax, 
        marker='o', 
        color=[colors[c] for c in plot_lines]
    )
    
    ax.set_title(title_text, fontsize=14, fontweight='bold')
    ax.set_xlabel('Observation Date', fontsize=12)
    ax.legend(title='Vehicle Type', loc=legend_loc)
    ax.tick_params(axis='x', rotation=45)
    
    ax.xaxis.set_major_formatter(date_formatter)
    ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
    ax.margins(x=0.05)


# --- Plot 1: 09:00–10:00 (Morning) ---
plot_traffic(
    df_morning, 
    axes[0], 
    'Figure 1: Daily Vehicle Traffic (09:00–10:00)', 
    'upper right'
)
axes[0].set_ylabel('Vehicle Count (Units)', fontsize=12)


# --- Plot 2: 14:00–15:00 (Afternoon) ---
plot_traffic(
    df_afternoon, 
    axes[1], 
    'Figure 2: Daily Vehicle Traffic (14:00–15:00)', 
    'upper left'
)
axes[1].set_ylabel('')

plt.suptitle(
    'Campus-wide Vehicle Traffic by Observation Day and Time Slot', 
    fontsize=16, 
    y=1.02, 
    fontweight='bold'
)
plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.savefig('daily_traffic_by_date_actual.png')
plt.show()
```

```python
df_daily_avg = df_filtered.groupby('Date')[['Total EV cars', 'Gasoline cars', 'Total cars']].mean().reset_index()

fig, ax = plt.subplots(figsize=(12, 7))
plt.style.use('seaborn-v0_8-whitegrid')

plot_lines = ['Total EV cars', 'Gasoline cars', 'Total cars']
colors = {'Total EV cars': 'skyblue', 'Gasoline cars': 'orange', 'Total cars': 'darkblue'}
date_formatter = mdates.DateFormatter('%m-%d')

df_daily_avg.set_index('Date')[plot_lines].plot(
    kind='line', 
    ax=ax, 
    marker='o', 
    color=[colors[c] for c in plot_lines], 
    linewidth=2
)

ax.set_title(
    'Average Daily Vehicle Traffic (Combined Morning & Afternoon Slots)', 
    fontsize=15, 
    fontweight='bold'
)
ax.set_ylabel('Average Vehicle Count (Units)', fontsize=12)
ax.set_xlabel('Observation Date', fontsize=12)
ax.legend(title='Vehicle Type', loc='upper right')
ax.tick_params(axis='x', rotation=45)
ax.grid(axis='y', linestyle='--', alpha=0.7)


ax.xaxis.set_major_formatter(date_formatter)
ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))

ax.margins(x=0.05)

plt.tight_layout()
plt.savefig('daily_traffic_average.png')
plt.show()
```

```python
df_combined = pd.concat([df_C1, df_J1], ignore_index=True)
df_combined['Date'] = pd.to_datetime(df_combined['Date'])
df_filtered = df_combined[df_combined['Total cars'] > 0].copy()

metrics = [
    'Total EV cars',
    'Gasoline cars',
    'Total cars',
    'Gasoline cars parked in active CP slot'
]
df_daily_avg = df_filtered.groupby(['Building', 'Date'])[metrics].mean().reset_index()

df_avg_C = df_daily_avg[df_daily_avg['Building'] == 'C'].sort_values(by='Date')
df_avg_J = df_daily_avg[df_daily_avg['Building'] == 'J'].sort_values(by='Date')
```

```python
metric_titles = {
    'Total EV cars': 'Total EV Cars',
    'Gasoline cars': 'Gasoline Cars',
    'Total cars': 'Total Vehicles',
    'Gasoline cars parked in active CP slot': 'Gasoline Cars in EV Charging Slot'
}

fig, axes = plt.subplots(2, 2, figsize=(18, 12), sharex=True)
axes = axes.flatten()

plt.style.use('seaborn-v0_8-whitegrid')
date_formatter = mdates.DateFormatter('%m-%d')

for i, metric in enumerate(metrics):
    ax = axes[i]
    
    # Plotting: C vs J
    ax.plot(df_avg_C['Date'], df_avg_C[metric], label='Building C', marker='o', linewidth=2, color='skyblue')
    ax.plot(df_avg_J['Date'], df_avg_J[metric], label='Building J', marker='x', linewidth=2, color='darkorange', linestyle='--')
    
    ax.set_title(f'Daily Average Traffic: {metric_titles[metric]}', fontsize=14, fontweight='bold')
    ax.set_ylabel('Average Vehicle Count (Units)', fontsize=11)
    ax.legend(loc='upper left')
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    
    if i >= 2:
        ax.set_xlabel('Observation Date', fontsize=12)
        ax.tick_params(axis='x', rotation=45)
        ax.xaxis.set_major_formatter(date_formatter)
        ax.xaxis.set_major_locator(mdates.DayLocator())

plt.suptitle('Comparison of Daily Average Vehicle Traffic between Building C and Building J', fontsize=16, y=1.02, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.savefig('daily_average_comparison_C_vs_J.png')
plt.show()
```
